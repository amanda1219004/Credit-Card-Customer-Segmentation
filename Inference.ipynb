# Import Libraries

```python
import numpy as np
import pandas as pd
import joblib
```

# Load Model dan Preprocessing

```python
imputer = joblib.load(
    "model_artifacts/imputer.joblib"
)

winsorizer = joblib.load(
   "model_artifacts/winsorizer.joblib" 
)

scaler = joblib.load(
    "model_artifacts/scaler.joblib"
)

pca = joblib.load(
    "model_artifacts/pca.joblib"
)

kmeans_model_fin = joblib.load(
    "model_artifacts/kmeans.joblib"
)
```

# Membuat Data Inference dalam Format Asli

```python
user = pd.DataFrame([{
    "CUST_ID": 9999,
    "BALANCE": 2000,
    "BALANCE_FREQUENCY": 0.5,
    "PURCHASES": 1000,
    "ONEOFF_PURCHASES": 600,
    "INSTALLMENTS_PURCHASES": 800,
    "CASH_ADVANCE": 500,
    "PURCHASES_FREQUENCY": 5,
    "ONEOFF_PURCHASES_FREQUENCY": 0.5,
    "PURCHASES_INSTALLMENTS_FREQUENCY": 0.7,
    "CASH_ADVANCE_FREQUENCY": 0.3,
    "CASH_ADVANCE_TRX": 3,
    "PURCHASES_TRX": 20,
    "CREDIT_LIMIT": 5000,
    "PAYMENTS": 800,
    "MINIMUM_PAYMENTS": 500,
    "PRC_FULL_PAYMENT": 0.2,
    "TENURE": 12
}])

user
```

```python
user_model = user[imputer.feature_names_in_].copy()
```

# Preprocessing dan Prediksi Cluster

```python
# 1. Ambil fitur yang digunakan saat training
user_model = user[imputer.feature_names_in_].copy()

# 2. Imputation
X_imputed = imputer.transform(user_model)

# Ubah kembali menjadi DataFrame
X_imputed = pd.DataFrame(
    X_imputed,
    columns=imputer.feature_names_in_,
    index=user_model.index
)

# 3. Winsorization
X_winsorized = winsorizer.transform(X_imputed)

# 4. Scaling
X_scaled = scaler.transform(X_winsorized)

# 5. PCA
X_pca = pca.transform(X_scaled)

# 6. Cek shape
print("Shape after imputer   :", X_imputed.shape)
print("Shape after winsorizer:", X_winsorized.shape)
print("Shape after scaler    :", X_scaled.shape)
print("Shape after PCA       :", X_pca.shape)

# 7. Prediction
prediction = kmeans_model_fin.predict(X_pca)

print("Predicted Cluster:", prediction[0])
```

# Kesimpulan Prediksi

Berdasarkan hasil inference, satu data customer berhasil melewati seluruh tahapan preprocessing tanpa mengubah jumlah fitur, yaitu dari 4 fitur menjadi tetap 4 fitur setelah imputation, winsorization, scaling, dan PCA. Model K-Means kemudian memprediksi customer tersebut masuk ke Cluster 0. Hasil ini menunjukkan bahwa data customer memiliki karakteristik yang lebih dekat dengan profil customer pada Cluster 0 berdasarkan pola yang telah dipelajari oleh model.
